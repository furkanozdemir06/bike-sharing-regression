The goal of this project is to predict the number of bikes rented per hour in a city bike-sharing system. Accurate predictions help operators optimize bike availability, improve customer satisfaction, and efficiently allocate resources across stations.

```python
#Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

import warnings

warnings.filterwarnings('ignore')
```

# EDA 

```python
test= pd.read_csv("test.csv")
train= pd.read_csv("train.csv")

print(train.shape)
print(test.shape)

train.head()
```

```python
train_len = len(train)
df = pd.concat([train, test], axis=0).reset_index(drop=True)
```

### Feature Engineering

```python
df["datetime"] = pd.to_datetime(df["datetime"])
```

```python
df["hour"] = df["datetime"].dt.hour

df["day"] = df["datetime"].dt.day

df["month"] = df["datetime"].dt.month

df['year'] = df['datetime'].dt.year

df['dayofweek'] = df['datetime'].dt.dayofweek
```

```python
df['season_label'] = df['season'].map({1: "Spring", 2: "Summer", 3: "Fall", 4: "Winter"})
```

```python
df['weather_label'] = df['weather'].map({1: "Clear", 2: "Cloudy", 3: "Rainy", 4: "Heavy_Rain"})
```

```python
df['day_name'] = df['dayofweek'].map({0:"Mon", 1:"Tue", 2:"Wed", 3:"Thu", 4:"Fri", 5:"Sat", 6:"Sun"})
```

```python
df['is_peak_hour'] = 0
df.loc[(df['workingday'] == 1) & (df['hour'].isin([8, 17, 18])), 'is_peak_hour'] = 1
```

```python
df = pd.get_dummies(df, columns=["season", "weather"], drop_first=False)
```

```python
train_final = df.iloc[:train_len].copy()
test_final = df.iloc[train_len:].copy()
```

```python
y_train = np.log1p(train["count"])
```

```python
drop_cols = [
    "datetime",
    "casual",
    "registered",
    "count",
    "season_label",
    "weather_label",
    "day_name"
]

x_train = train_final.drop(columns=drop_cols, errors="ignore")
x_test = test_final.drop(columns=drop_cols, errors="ignore")
```

### Modeling

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model = RandomForestRegressor(n_estimators=500, max_depth=15, random_state=42, n_jobs=-1)
```

```python
model.fit(x_train, y_train)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

x_tr, x_val, y_tr, y_val = train_test_split(
    x_train, y_train, test_size=0.2, random_state=42
)

model.fit(x_tr, y_tr)

val_pred = model.predict(x_val)

val_pred_real = np.expm1(val_pred)
y_val_real = np.expm1(y_val)

r2 = r2_score(y_val_real, val_pred_real)
rmse = np.sqrt(mean_squared_error(y_val_real, val_pred_real))
mae = mean_absolute_error(y_val_real, val_pred_real)

print("R2 Score:", r2)
print("RMSE:", rmse)
print("MAE:", mae)
```

```python
#Prediction
pred_log = model.predict(x_test)

pred = np.expm1(pred_log)
pred = np.maximum(pred, 0)
```

```python
#Submission
submission = pd.DataFrame({"datetime": test["datetime"],"count": pred})

submission.head()
```

```python
submission.to_csv("submission.csv", index=False)
```

By applying log transformation and time-based feature engineering, the Random Forest model achieved a strong $R^2$ score of ~0.948 (RMSE: 41.51, MAE: 25.58) and successfully exported final predictions to `submission.csv`.
